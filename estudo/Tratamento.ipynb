```python
import pandas as pd
```

```python
df_ini = pd.read_csv("../data/dataset_01_parceiroA.csv")
df_ini
```

```python
df = df_ini.copy(deep=True)
df.head()
```

```python
df.isna().sum()
```

```python
# Tratando nome das colunas

df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")
df.head()
```

```python
# Tratando colunas inteiras
df["comissão"] = df["comissão"].str.replace("$", "").str.replace(".", "").str.strip().str.replace("R", "")
df["cashback"] = df["cashback"].str.replace("$", "").str.replace(".", "").str.strip().str.replace("R", "")
df["vendas_totais"] = df["vendas_totais"].str.replace("$", "").str.replace(".", "").str.strip().str.replace("R", "")

# Tratando colunas str
df["grupos_de_usuários"] = df["grupos_de_usuários"].str.lower().str.strip().str.replace(" ", "_")
df["parceiro"] = df["parceiro"].str.lower().str.strip().str.replace(" ", "_")

df.head()
```

```python
# Transformando numeros em int
df["comissão"] = pd.to_numeric(df["comissão"], errors="coerce")
df["cashback"] = pd.to_numeric(df["cashback"], errors="coerce")
df["vendas_totais"] = pd.to_numeric(df["vendas_totais"], errors="coerce")

df.dtypes
```

```python
df["data"] = pd.to_datetime(df["data"], errors="coerce")
df.head()
```
