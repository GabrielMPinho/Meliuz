```python
import sys
import pandas as pd
sys.path.append("../src")

from limpeza import data_cleaning

```

```python
df = data_cleaning("../data/dataset_03_parceiroC.csv")
df.head()
```

```python
df["grupos_de_usuários"].unique()
```

```python
df_grupo1 = df[df["grupos_de_usuários"] == "grupo_1"]
df_grupo2 = df[df["grupos_de_usuários"] == "grupo_2"]
df_grupo3 = df[df["grupos_de_usuários"] == "grupo_3"]
```

## Lucro Líquido

```python
comissao_grp1 = df_grupo1["comissão"].sum()
cashback_grp1 = df_grupo1["cashback"].sum()

print(comissao_grp1, cashback_grp1)

lucro_liquido_grp1 = comissao_grp1 - cashback_grp1
```

```python
comissao_grp2 = df_grupo2["comissão"].sum()
cashback_grp2 = df_grupo2["cashback"].sum()

print(comissao_grp2, cashback_grp2)

lucro_liquido_grp2 = comissao_grp2 - cashback_grp2
```

```python
comissao_grp3 = df_grupo3["comissão"].sum()
cashback_grp3 = df_grupo3["cashback"].sum()

print(comissao_grp3, cashback_grp3)

lucro_liquido_grp3 = comissao_grp3 - cashback_grp3
```

```python
print(f"Lucro líquido do grupo 1: {lucro_liquido_grp1}")
print(f"Lucro líquido do grupo 2: {lucro_liquido_grp2}")
print(f"Lucro líquido do grupo 3: {lucro_liquido_grp3}")
```

### ROI

```python
df.head()
```

```python
roi_grp1 = lucro_liquido_grp1 / cashback_grp1
roi_grp2 = lucro_liquido_grp2 / cashback_grp2
roi_grp3 = lucro_liquido_grp3 / cashback_grp3

print(f"ROI grupo 1: {roi_grp1:.2f}\nROI grupo 2: {roi_grp2:.2f}\nROI grupo 3: {roi_grp3:.2f}")
```

## Lucro médio por comprador

```python
total_compradores_grp1 = df_grupo1["compradores"].sum()
lucro_medio_comprador_grp1 = lucro_liquido_grp1 / total_compradores_grp1

total_compradores_grp2 = df_grupo2["compradores"].sum()
lucro_medio_comprador_grp2 = lucro_liquido_grp2 / total_compradores_grp2

total_compradores_grp3 = df_grupo3["compradores"].sum()
lucro_medio_comprador_grp3 = lucro_liquido_grp3 / total_compradores_grp3

print(f"Lucro médio por comprador grupo 1: {lucro_medio_comprador_grp1:.2f}\nLucro médio por comprador grupo 2: {lucro_medio_comprador_grp2:.2f}\nLucro médio por comprador grupo 3: {lucro_medio_comprador_grp3:.2f}")
```

## Fluxo simplificado

```python
df_agrupado = df.groupby("grupos_de_usuários").agg({
    "comissão": "sum",
    "cashback": "sum",
    "compradores": "sum",
    "vendas_totais": "sum"
}
)
df_agrupado 
```

```python
df_kpis = (
    df.groupby("grupos_de_usuários")
    .agg({
        "comissão": "sum",
        "cashback": "sum",
        "compradores": "sum",
        "vendas_totais": "sum",
    })
    .reset_index()
)

df_kpis["lucro_liquido_total"] = df_kpis["comissão"] - df_kpis["cashback"]
df_kpis["roi"] = df_kpis["lucro_liquido_total"] / df_kpis["cashback"]
df_kpis["lucro_por_comprador"] = df_kpis["lucro_liquido_total"] / df_kpis["compradores"]
df_kpis["gmv"] = df_kpis["vendas_totais"]

df_kpis
```
